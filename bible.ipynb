# Bibliography

Данный обзор структурирует литературу по трем основным категориям: фундаментальные учебники и монографии (Books), научные статьи (Articles) и прикладные руководства (Practical Guides). Цель обзора — выделить ключевые идеи, методы и контекст каждого источника применительно к задаче прогнозирования продаж.

## Books

В эту категорию включены источники, закладывающие теоретическую базу для понимания методов прогнозирования, от классической статистики до современных подходов

### 1.1. Классические и статистические методы

*   **Box, George, and Jenkins, GM. "Analysis: Forecasting and control." (1976).**
    *   **Контекст:** Это оригинальная работа, представившая миру методологию ARIMA (Autoregressive Integrated Moving Average). Бокс-Дженкинс — это золотой стандарт для построения моделей ARIMA, который включает итеративный процесс идентификации модели, оценки параметров и проверки адекватности.
    *   **Значение для диплома:** Данный источник является первоосновой для понимания классического подхода прогнозирования продаж в условиях сезонности и ярко выраженного тренда

*   **Лукашин, Юрий Павлович. "Адаптивные методы краткосрочного прогнозирования временных рядов." Финансы и статистика, 2003.**
    *   **Контекст:** Классический русскоязычный труд, посвященный адаптивным методам (экспоненциальное сглаживание, модели Брауна, Хольта и др.). Эти методы особенно эффективны для краткосрочного прогнозирования в условиях нестационарности рядов.
    *   **Значение для диплома:** Работа важна для раздела, посвященного классическим методам, особенно если прогнозирование продаж является краткосрочным. Адаптивные методы часто служат надежным бенчмарком для сравнения с более сложными моделями.

*   **Popova, Irina, and Valeriy Kovalev. "Анализ временных рядов. Учебник для вузов." LitRes, 2023.**
    *   **Контекст:** Современный российский учебник, который, вероятно, обобщает как классическую теорию, так и современные практики анализа временных рядов.
    *   **Значение для диплома:** Может служить хорошей базой для написания теоретической части, посвященной основным понятиям: стационарность, автокорреляция, декомпозиция ряда (тренд, сезонность, остатки).

*   **Lawton, Richard. "Time Series Analysis and its Applications." (рецензия на книгу Shumway и Stoffer).**
    *   **Контекст:** Это рецензия на один из самых авторитетных учебников по анализу временных рядов (Shumway and Stoffer). Оригинальная книга славится балансом между теорией и практическими примерами (часто на R, но идеи универсальны).
    *   **Значение для диплома:** Указывает на глубокую теоретическую проработку вопроса

### 1.2. Современные методы и машинное обучение

* **Hyndman, R.J., and Athanasopoulos, G. "Forecasting: principles and practice." OTexts, 2018.**
    *   **Контекст:** Это, пожалуй, самая популярная и доступная современная книга по прогнозированию. Она охватывает как классику (экспоненциальное сглаживание, ARIMA), так и более сложные темы (динамическая регрессия, иерархическое прогнозирование). Авторы — признанные авторитеты в области прогнозирования (Hyndman — автор пакета `forecast` для R).
    *   **Значение для диплома:** Обязательна к использованию для описания методологии, метрик ошибок (MAE, RMSE, MASE), визуализации и выбора моделей.

* **Huang, Changquan, and Alla Petukhina. "Applied time series analysis and forecasting with Python." Springer, 2022.**
    *   **Контекст:** Современное прикладное руководство, ориентированное на Python. Связывает теоретические концепции с их практической реализацией с помощью популярных библиотек (pandas, statsmodels, scikit-learn).
    *   **Значение для диплома:** Поможет с описанием инструментов, используемых для реализации моделей.

* **Joseph, Manu, and J. Tackes. "Modern time series forecasting with Python." Packt Publishing Ltd, 2022.**
    *   **Контекст:** Еще одно практическое руководство по Python, фокусирующееся на современных подходах, включая глубокое обучение (LSTM, CNN) для временных рядов.
    *   **Значение для диплома:** Полезен для раздела, посвященного "машинным методам"

* **Siegel, Eric. "Predictive analytics: The power to predict who will click, buy, lie, or die." John Wiley & Sons, 2013.**
    *   **Контекст:** Книга, скорее, про общую философию и применение предиктивной аналитики в бизнесе, а не про технические детали.
    *   **Значение для диплома:** Может быть использована во введении или заключении для обоснования практической ценности прогнозирования продаж для бизнеса ("почему это важно").

* **Sario, Azhar Ul Haque. "Sales Forecasting: Data Science Models." Azhar ul Haque Sario, 2023.**
    *   **Контекст:** Узкоспециализированная работа, напрямую посвященная теме прогнозирования продаж. Должна содержать обзор различных моделей данных именно в контексте Retail- или B2B-продаж.
    *   **Значение для диплома:**  Позволит описать специфику данных о продажах (сезонность, праздники, промо-акции) и то, как она учитывается в моделях.

## 2. Научные статьи (Articles)
*В этой категории собраны исследования, представляющие собой либо классические примеры применения методов, либо современные разработки и сравнения подходов.*

### 2.1. Исследования классических методов (ARIMA/SARIMA)
*   **Ediger, Volkan Ş., and Sertaç Akar. "ARIMA forecasting of primary energy demand by fuel in Turkey." Energy Policy 35.3 (2007): 1701-1708.**
    *   **Контекст:** Классическая прикладная статья, демонстрирующая мощь ARIMA и SARIMA (сезонного ARIMA) для прогнозирования в энергетике. Авторы подчеркивают, что в условиях нестабильности экономических параметров, чистые временные ряды (time-series) могут давать более надежные прогнозы, чем эконометрические модели.
    *   **Значение для диплома:**  Показывает, как грамотно применять и интерпретировать SARIMA. Вывод о том, что агрегированный прогноз надежнее суммы индивидуальных, может быть полезен для прогнозирования продажи для нескольких продуктов.

### 2.2. Сравнительные исследования и современные тенденции
*   **Makridakis, Spyros, Evangelos Spiliotis, and Vassilios Assimakopoulos. "The M4 Competition: Results, findings, conclusion and way forward." International Journal of forecasting 34.4 (2018): 802-808.**
    *   **Контекст:** Результаты знаменитого соревнования M4 по прогнозированию, которое возглавляет патриарх прогнозирования Спирос Макридакис. Это масштабное сравнение десятков методов — от простых статистических до сложных "черных ящиков" машинного обучения.
    *   **Значение для диплома:** Ключевой источник. Главный вывод M4: комбинированные (ensemble) и гибридные методы часто работают лучше всего, а простые статистические модели (например, комбинация экспоненциального сглаживания) удивительно конкурентоспособны. Это отличная база для обсуждения выбора методов в работе.

*   **Elsayed, Shereen, et al. "Do we really need deep learning models for time series forecasting?." arXiv preprint arXiv:2101.02118 (2021).**
    *   **Контекст:** Статья, задающая провокационный вопрос о необходимости глубокого обучения. Авторы проводят эксперименты и сравнивают сложные нейросети с более простыми методами (например, линейной регрессией на лагах, LightGBM).
    *   **Значение для диплома:** Важна для критического анализа. Результаты часто показывают, что правильно настроенные "не-глубокие" модели машинного обучения (бустинг, случайный лес) или даже классика могут не уступать LSTM и Transformer-архитектурам. Это поможет аргументировать выбор методов и не переусложнять модель без необходимости.

*   **Bontempi, Gianluca, Souhaib Ben Taieb, and Yann-Aël Le Borgne. "Machine learning strategies for time series forecasting." Business Information Systems (2013).**
    *   **Контекст:** Глава в сборнике, посвященная стратегиям машинного обучения. Должна подробно разбирать такие концепции, как рекуррентное (recursive) и прямые (direct) стратегии многошагового прогнозирования, а также методы на основе скользящего окна.
    *   **Значение для диплома:** Технически важный источник для понимания того, как именно алгоритмы машинного обучения (регрессия, SVM, деревья) адаптируются для задачи прогнозирования временных рядов.

### 2.3. Прикладные исследования в ритейле и смежных областях
*   **Sajawal, Muhammad, et al. "Predictive Analysis of Retail Sales Forecasting using Machine Learning Techniques." (2023).**
    *   **Контекст:** Непосредственно прикладное исследование по прогнозированию розничных продаж с помощью машинного обучения.
    *   **Значение для диплома:** Прямой аналог работы. Позволит сослаться на опыт других исследователей в той же предметной области, сравнить их подходы и результаты с собственными.

*   **Qureshi, Nameer, et al. "Demand Forecasting in Supply Chain Management for Rossmann Stores Using Weather Enhanced Deep Learning Model." IEEE Access (2024).**
    *   **Контекст:** Очень современное исследование (2024 год) на примере известных магазинов Rossmann. Особенность — использование внешних факторов (weather), что выводит модель за рамки чистого временного ряда.
    *   **Значение для диплома:** Отличный пример расширения модели (feature engineering) за счет добавления экзогенных переменных.  Также демонстрирует применение Deep Learning в задаче прогноза спроса.

*   **Taylor, Sean J., and Benjamin Letham. "Forecasting at scale." The American Statistician 72.1 (2018): 37-45.**
    *   **Контекст:** Статья, представляющая модель Prophet, разработанную в Facebook. Prophet — это аддитивная регрессионная модель, специально созданная для прогнозирования временных рядов с сильной сезонностью и трендом, а также с возможностью ручной корректировки экспертом.
    *   **Значение для диплома:** Эта статья "материнская" публикация модели Prophet. Она описывает мотивацию создания модели, ее компоненты (тренд, сезонность, праздники) и преимущества для бизнес-задач.


## 3. Выводы и связь с темой диплома

Анализ представленной литературы позволяет сформировать прочную базу для дипломной работы:

1.  **Теоретический фундамент:** Источники, такие как **Box & Jenkins (1976)**, **Лукашин (2003)** и **Hyndman & Athanasopoulos (2018)**, дают исчерпывающее описание классических методов (ARIMA/SARIMA, экспоненциальное сглаживание), которые будут выступать в роли базовых моделей (benchmarks).

2.  **Современная методология:** Книги **Huang & Petukhina (2022)** и **Joseph (2022)** обеспечат практическую реализацию этих методов на Python. Работы **Bontempi et al. (2013)** и **Elsayed et al. (2021)** помогут грамотно адаптировать алгоритмы машинного обучения для временных рядов и критически оценить необходимость их применения.

3.  **Прикладной контекст:** Исследования **Sajawal et al. (2023)** и **Qureshi et al. (2024)** показывают, как именно задачи прогнозирования решаются в сфере розничной торговли и управления цепочками поставок. Статья **Ediger & Akar (2007)** служит примером классического подхода SARIMA.

4.  **Обоснование выбора методов:** Выводы соревнования **M4 Competition (Makridakis et al., 2018)** и провокационное исследование **Elsayed et al. (2021)** дают веские аргументы для сравнения "классики" и "машинного обучения", а также подчеркивают ценность комбинированных подходов. Модель **Prophet (Taylor & Letham, 2018)** представляет собой современный гибридный метод, учитывающий бизнес-специфику (праздники, изменения тренда).

Таким образом, данная подборка литературы охватывает все ключевые аспекты темы: от исторических корней и классических методов до самых современных дискуссий о роли глубокого обучения и специфике прогнозирования в ритейле.
